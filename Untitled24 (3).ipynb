```python
import pandas as pd

def rank_weekend_destinations(source_city, top_n=5):
    # Load dataset
    df = pd.read_csv("Top Indian Places to Visit.csv")

    # Clean column names
    df.columns = df.columns.str.strip()

    # Remove places from the same city
    df = df[df["City"] != source_city]

    # Handle missing values
    df["Google review rating"] = df["Google review rating"].fillna(0)
    df["Number of google review in lakhs"] = df["Number of google review in lakhs"].fillna(0)
    df["time needed to visit in hrs"] = df["time needed to visit in hrs"].fillna(
        df["time needed to visit in hrs"].mean()
    )

    # Normalize scores
    df["rating_score"] = df["Google review rating"] / df["Google review rating"].max()
    df["popularity_score"] = (
        df["Number of google review in lakhs"] /
        df["Number of google review in lakhs"].max()
    )
    df["time_score"] = 1 - (
        df["time needed to visit in hrs"] /
        df["time needed to visit in hrs"].max()
    )

    # Final score
    df["final_score"] = (
        0.45 * df["rating_score"] +
        0.35 * df["popularity_score"] +
        0.20 * df["time_score"]
    )

    # Sort and return top results
    result = df.sort_values(by="final_score", ascending=False)

    return result[[
        "Name",
        "City",
        "State",
        "Type",
        "Google review rating",
        "Number of google review in lakhs",
        "time needed to visit in hrs",
        "final_score"
    ]].head(top_n)


if __name__ == "__main__":
    cities = ["Kolkata", "Delhi", "Bangalore"]

    for city in cities:
        print(f"\nTop Weekend Destinations (Excluding {city})")
        print(rank_weekend_destinations(city))
```
