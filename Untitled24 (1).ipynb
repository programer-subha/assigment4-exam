```python
import pandas as pd
import math

# Haversine formula to calculate distance
def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)
    dphi = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)

    a = math.sin(dphi/2)**2 + \
        math.cos(phi1) * math.cos(phi2) * math.sin(dlambda/2)**2

    return 2 * R * math.atan2(math.sqrt(a), math.sqrt(1 - a))


def rank_destinations(source_city, source_lat, source_lon, top_n=5):
    df = pd.read_csv("travel_places.csv")

    # Calculate distance
    df["distance_km"] = df.apply(
        lambda row: haversine(
            source_lat, source_lon, row["latitude"], row["longitude"]
        ),
        axis=1
    )

    # Normalize distance (closer is better)
    df["distance_score"] = 1 - (df["distance_km"] / df["distance_km"].max())

    # Normalize popularity
    df["popularity_score"] = df["popularity"] / df["popularity"].max()

    # Normalize rating
    df["rating_score"] = df["rating"] / df["rating"].max()

    # Final ranking score
    df["final_score"] = (
        0.4 * df["rating_score"] +
        0.35 * df["popularity_score"] +
        0.25 * df["distance_score"]
    )

    ranked = df.sort_values(by="final_score", ascending=False)

    return ranked[[
        "place", "state", "distance_km", "rating", "popularity", "final_score"
    ]].head(top_n)


if __name__ == "__main__":
    # Example source cities
    sources = {
        "Kolkata": (22.5726, 88.3639),
        "Delhi": (28.6139, 77.2090),
        "Bangalore": (12.9716, 77.5946)
    }

    for city, coords in sources.items():
        print(f"\nTop Weekend Destinations from {city}")
        print(rank_destinations(city, coords[0], coords[1]))
```

```python
import pandas as pd

def rank_weekend_destinations(source_city, top_n=5):
    # Load dataset
    df = pd.read_csv("Top Indian Places to Visit.csv")

    # Clean column names
    df.columns = df.columns.str.strip()

    # Remove places from the same city
    df = df[df["City"] != source_city]

    # Handle missing values
    df["Google review rating"] = df["Google review rating"].fillna(0)
    df["Number of google review in lakhs"] = df["Number of google review in lakhs"].fillna(0)
    df["time needed to visit in hrs"] = df["time needed to visit in hrs"].fillna(
        df["time needed to visit in hrs"].mean()
    )

    # Normalize scores
    df["rating_score"] = df["Google review rating"] / df["Google review rating"].max()
    df["popularity_score"] = (
        df["Number of google review in lakhs"] /
        df["Number of google review in lakhs"].max()
    )
    df["time_score"] = 1 - (
        df["time needed to visit in hrs"] /
        df["time needed to visit in hrs"].max()
    )

    # Final score
    df["final_score"] = (
        0.45 * df["rating_score"] +
        0.35 * df["popularity_score"] +
        0.20 * df["time_score"]
    )

    # Sort and return top results
    result = df.sort_values(by="final_score", ascending=False)

    return result[[
        "Name",
        "City",
        "State",
        "Type",
        "Google review rating",
        "Number of google review in lakhs",
        "time needed to visit in hrs",
        "final_score"
    ]].head(top_n)


if __name__ == "__main__":
    cities = ["Kolkata", "Delhi", "Bangalore"]

    for city in cities:
        print(f"\nTop Weekend Destinations (Excluding {city})")
        print(rank_weekend_destinations(city))
```
